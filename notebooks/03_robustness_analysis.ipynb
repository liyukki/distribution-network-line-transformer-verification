# 鲁棒性小矩阵分析

在 1 天小规模配置上运行 8 个案例（缺失率 2 水平 × 2 种子 + 2 个消融 × 2 种子），展示聚合结果。完整 130 案例矩阵请运行 python -m ltverify experiments --config configs/robustness.yaml。

```python
from pathlib import Path


def repo_root() -> Path:
    for candidate in (Path.cwd(), Path.cwd().parent, Path.cwd().parent.parent):
        if (candidate / "configs" / "default.yaml").exists():
            return candidate
    raise FileNotFoundError("未找到仓库根目录")

ROOT = repo_root()
```

```python
from ltverify.experiments import run_experiments

runs = ROOT / "runs"
runs.mkdir(exist_ok=True)
robustness_yaml = runs / "notebook_robustness.yaml"
robustness_yaml.write_text("base_config: " + (ROOT / "tests/fixtures/small_config.yaml").as_posix() + "\n" + "experiments:\n" + "  missing_rate: [0.0, 0.05]\n" + "ablation_features: [full, without_difference]\n" + "seeds: [42, 43]\n", encoding="utf-8")
output_dir = runs / "notebook-experiments"
output_dir.mkdir(exist_ok=True)
run_experiments(robustness_yaml, output_dir)
print("实验目录:", output_dir.resolve())
```

```python
import pandas as pd

summary = pd.read_csv(output_dir / "robustness_summary.csv")
summary[["case_id", "family", "value", "seed", "f1", "top1_correction_rate", "status"]]
```

```python
import plotly.graph_objects as go

families = summary["family"].unique().tolist()
figure = go.Figure()
for family in families:
    rows = summary[summary["family"] == family]
    figure.add_trace(go.Bar(x=rows["value"].astype(str) + "-seed" + rows["seed"].astype(str), y=rows["f1"], name=family))
figure.update_layout(title="03 鲁棒性小矩阵 F1", yaxis_title="F1")
output = ROOT / "reports/figures/03_robustness_analysis.html"
output.parent.mkdir(parents=True, exist_ok=True)
figure.write_html(output)
print("图已保存:", output.resolve())
```
